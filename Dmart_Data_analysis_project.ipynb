#### **Q:-How many unique and repeated customers are there in the dataset, and what is their percentage distribution?**

```python
# Customer Overview: Unique & Repeated
total_records = len(df)
print("Total number of purchase records:", total_records)

unique = len(df['FullName'].unique())
print("Total Unique Customers:", unique)

repeated = total_records - unique
print("Total Repeated Customers:", repeated)

# ➡️ Corrected Pie Chart
plt.figure(figsize=(6,6))
plt.pie([float(unique), float(repeated)], 
        labels=['Unique Customers', 'Repeated Customers'], 
        autopct=lambda p: '{:.1f}%'.format(p) if p > 0 else '',  # avoid 0.0%
        colors=['skyblue', 'orange'], startangle=90)
plt.title("Unique vs Repeated Customers")
plt.show()
```

#### **Q:-Who is the most frequent customer and the least frequent customer in the dataset, how many times did they visit, and what is their customer type?**

#### **Q:-Which month had the maximum and minimum number of unique sales days, how many sales happened in those months, and in which year?**

```python

# Make sure Date is in datetime format
df["Date"] = pd.to_datetime(df["Date"])

# Extract Year, Month name, Day
df["Year"] = df["Date"].dt.year
df["Month_Name"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day

# Count unique days + total sales per month
days_per_month = df.groupby(['Year','Month_Name']).agg(
    UniqueDays=('Day','nunique'),          # how many different days
    TotalSales=('ProductCategory','count') # how many total sales
).reset_index()

# Find month with max days
max_days = days_per_month.loc[days_per_month['UniqueDays'].idxmax()]

# Find month with min days
min_days = days_per_month.loc[days_per_month['UniqueDays'].idxmin()]

print("Month with maximum :", max_days['Month_Name'], 
      "-", max_days['UniqueDays'], ",", 
      max_days['TotalSales'], "sales, Year:", max_days['Year'])

print("Month with minimum :", min_days['Month_Name'], 
      "-", min_days['UniqueDays'], ",", 
      min_days['TotalSales'], "sales, Year:", min_days['Year'])


```

#### **Q:-For each year, which product category had the maximum sales and which had the minimum sales?**

```python
# Convert Date column to datetime (if not already)
df["Date"] = pd.to_datetime(df["Date"])

# Create a new Year column from Date
df["Year"] = df["Date"].dt.year

# Group by Year + ProductCategory
year_category = df.groupby(['Year', 'ProductCategory']).size().reset_index(name='Count')

# For each Year, pick the ProductCategory with max sales
max_per_year = year_category.loc[year_category.groupby('Year')['Count'].idxmax()]

# For each Year, pick the ProductCategory with min sales
min_per_year = year_category.loc[year_category.groupby('Year')['Count'].idxmin()]

print("Max sales per year:")
print(max_per_year[['Year','ProductCategory','Count']])

print("\nMin sales per year:")
print(min_per_year[['Year','ProductCategory','Count']])

# ➡️ Chart: Max & Min Sales per Year
plt.figure(figsize=(10,6))
plt.bar(max_per_year['Year'] - 0.1, max_per_year['Count'], width=0.2, label='Max Sales', color='green')
plt.bar(min_per_year['Year'] + 0.1, min_per_year['Count'], width=0.2, label='Min Sales', color='red')

# Add labels for bars
for i, y in enumerate(max_per_year['Count']):
    plt.text(max_per_year['Year'].iloc[i] - 0.1, y + 0.5, y, ha='center', va='bottom')
for i, y in enumerate(min_per_year['Count']):
    plt.text(min_per_year['Year'].iloc[i] + 0.1, y + 0.5, y, ha='center', va='bottom')

plt.xticks(max_per_year['Year'])
plt.xlabel("Year")
plt.ylabel("Number of Sales")
plt.title("Max and Min Product Category Sales per Year")
plt.legend()
plt.show()
```

#### **Q:-For each month of each year, which product category had the maximum sales, and what were the counts?**

```python

# Make sure Date is datetime
df["Date"] = pd.to_datetime(df["Date"])

# Extract Year and Month_Name
df["Year"] = df["Date"].dt.year
df["Month_Name"] = df["Date"].dt.month_name()

# Remove year 2025
df_filtered = df[df['Year'] != 2025]

# Group by Year, Month_Name, ProductCategory
month_year_category = df_filtered.groupby(
    ['Year','Month_Name','ProductCategory'], observed=True
).size().reset_index(name='Count')

# For each Year+Month, pick the ProductCategory with max sales
max_per_month_year = month_year_category.loc[
    month_year_category.groupby(['Year','Month_Name'])['Count'].idxmax()
]

# Sort by Year and Month order (numeric month helps)
df["Month_Num"] = df["Date"].dt.month
max_per_month_year = max_per_month_year.merge(
    df[['Year','Month_Name','Month_Num']].drop_duplicates(),
    on=['Year','Month_Name'],
    how='left'
).sort_values(['Year','Month_Num'])

print(max_per_month_year[['Year','Month_Name','ProductCategory','Count']])
```

#### **Q:-Which product category has the highest sales in each time range (Morning, Afternoon, Evening, Night)?**

```python



# Example: if Time column is like "14:35:20"
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S")

# Create TimeRange column
def get_time_range(time):
    hour = time.hour
    if 6 <= hour < 12:
        return "Morning (6AM-12PM)"
    elif 12 <= hour < 18:
        return "Afternoon (12PM-6PM)"
    elif 18 <= hour < 24:
        return "Evening (6PM-12AM)"
    else:
        return "Night (12AM-6AM)"

df["TimeRange"] = df["Time"].apply(get_time_range)

# Now check which category sells more in each range
time_sales = df.groupby(["TimeRange","ProductCategory"]).size().reset_index(name="Total")

# Find top-selling category for each time range
max_per_timerange = time_sales.loc[time_sales.groupby("TimeRange")["Total"].idxmax()]

print("Top-selling category in each time range:")
print(max_per_timerange)
```

#### **Q:-For each day, which time range had the highest sales, and what was the total quantity sold?**

```python
# Make sure Date and Time are in correct format
df["Date"] = pd.to_datetime(df["Date"])
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S")

# Extract Year, Month, Day, DayName, Hour
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()
df["Hour"] = df["Time"].dt.hour

# Define time ranges (Morning, Afternoon, Evening, Night)
def time_period(hour):
    if 5 <= hour < 12:
        return "Morning (5AM–12PM)"
    elif 12 <= hour < 17:
        return "Afternoon (12PM–5PM)"
    elif 17 <= hour < 21:
        return "Evening (5PM–9PM)"
    else:
        return "Night (9PM–5AM)"

df["Time_Range"] = df["Hour"].apply(time_period)

# Group by Year, Month, Date, Day_Name, Time_Range
range_sales = df.groupby(
    ["Year", "Month", "Date", "Day_Name", "Time_Range"]
)["Quantity"].sum().reset_index()

# Find peak time range for each Year + Month + Date
peak_ranges = range_sales.loc[
    range_sales.groupby(["Year", "Month", "Date"])["Quantity"].idxmax()
]

print("Peak Sales by Year, Month, Date, Day and Time Range:")
print(peak_ranges[["Year", "Month", "Date", "Day_Name", "Time_Range", "Quantity"]])
```

#### **Q:-For each time range (Morning, Afternoon, Evening, Night), which customer type purchased the highest quantity, and what were the quantities?**

```python
# Make sure Date and Time are in correct format
df["Date"] = pd.to_datetime(df["Date"])
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M:%S")

# Extract Hour
df["Hour"] = df["Time"].dt.hour

# Define time ranges
def time_period(hour):
    if 5 <= hour < 12:
        return "Morning (5AM–12PM)"
    elif 12 <= hour < 17:
        return "Afternoon (12PM–5PM)"
    elif 17 <= hour < 21:
        return "Evening (5PM–9PM)"
    else:
        return "Night (9PM–5AM)"

df["Time_Range"] = df["Hour"].apply(time_period)

# Group by CustomerType and Time_Range, sum Quantity
customer_time = df.groupby(["CustomerType", "Time_Range"])["Quantity"].sum().reset_index()

# For each Time_Range, find the CustomerType with max Quantity
top_customer_time = customer_time.loc[
    customer_time.groupby("Time_Range")["Quantity"].idxmax()
]

print("Customer Type with highest quantity in each Time Range:")
print(top_customer_time[["Time_Range", "CustomerType", "Quantity"]])

# ➡️ Chart: Top CustomerType per Time Range
plt.figure(figsize=(8,5))
bars = plt.bar(top_customer_time['Time_Range'], top_customer_time['Quantity'], 
               color='skyblue', edgecolor='black')

plt.title("Top CustomerType by Quantity in Each Time Range")
plt.xlabel("Time Range")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)

# Add values on top of bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, int(yval), ha='center', va='bottom')

plt.show()
```
